```python
# Definición de la clase Árbitro
class Arbitro:
    # El método __init__ define los ATRIBUTOS de la clase
    def __init__(self, nombre, categoria, experiencia):
        self.nombre = nombre            # Atributo 1
        self.categoria = categoria      # Atributo 2
        self.experiencia = experiencia  # Atributo 3

    # FUNCIÓN 1: pitar_falta()
    def pitar_falta(self):
        print(f"[{self.nombre}]: ¡PIIIIII! Se detiene el juego por una infracción.")

    # FUNCIÓN 2: sacar_tarjeta()
    def sacar_tarjeta(self, color, jugador):
        print(f"[{self.nombre}]: Muestra tarjeta {color} al jugador {jugador}.")

    # FUNCIÓN 3: revisar_VAR()
    def revisar_VAR(self, jugada):
        print(f"[{self.nombre}]: Se detiene el juego. Revisando '{jugada}' en el VAR...")
        print(f"[{self.nombre}]: Decisión tomada tras revisar la pantalla.")

# --- CREACIÓN DE OBJETOS ---

# Objeto 1: Arbitro1
arbitro1 = Arbitro("Pierluigi Collina", "Internacional", 25)

# Objeto 2: Arbitro2
arbitro2 = Arbitro("Mariana Martínez", "Primera División", 8)

# --- EJECUCIÓN DE ACCIONES (Llamada a funciones) ---

print("--- ACCIONES DEL ÁRBITRO 1 ---")
print(f"Árbitro: {arbitro1.nombre} | Categoría: {arbitro1.categoria}")
arbitro1.pitar_falta()
arbitro1.sacar_tarjeta("Amarilla", "Número 10")
print("-" * 30)

print("--- ACCIONES DEL ÁRBITRO 2 ---")
print(f"Árbitro: {arbitro2.nombre} | Años de experiencia: {arbitro2.experiencia}")
arbitro2.revisar_VAR("Posible fuera de juego")
arbitro2.sacar_tarjeta("Roja", "Portero")

print("programa finalizado")
print("creador de software Nicole Robles NL = 46")
```
